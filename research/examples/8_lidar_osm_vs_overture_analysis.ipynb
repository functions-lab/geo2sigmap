```python
import json
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
```

```python
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
```

```python
STATS_FILE = Path("./stats_progress.json")

with open(STATS_FILE, "r") as f:
        data = json.load(f)

relative_to_lidar_osm = data.get("relative_to_lidar_osm")
lidar_osm_hag_perc = data.get("lidar_osm_hag_perc")
lidar_osm_building_height_perc = data.get("lidar_osm_building_height_perc")
lidar_osm_height_perc = data.get("lidar_osm_height_perc")
lidar_osm_building_levels_perc = data.get("lidar_osm_building_levels_perc")
lidar_osm_levels_perc = data.get("lidar_osm_levels_perc")
lidar_osm_random_fallback_perc = data.get("lidar_osm_random_fallback_perc")
overture_height_perc = data.get("overture_height_perc")
overture_num_floors_perc = data.get("overture_num_floors_perc")
overture_random_fallback_perc = data.get("overture_random_fallback_perc")
overall_mean_abs_diff = data.get("overall_mean_abs_diff")
overall_max_abs_diff = data.get("overall_max_abs_diff")
lidar_outside_explicit_overture_height = data.get("lidar_outside_explicit_overture_height")
lidar_osm_scene_gen_errors = data.get("lidar_osm_scene_gen_errors")
overture_scene_gen_errors = data.get("overture_scene_gen_errors")
```

```python
print(len(relative_to_lidar_osm), relative_to_lidar_osm)

print(len(lidar_osm_hag_perc), lidar_osm_hag_perc)
print(len(lidar_osm_building_height_perc), lidar_osm_building_height_perc)
print(len(lidar_osm_height_perc), lidar_osm_height_perc)
print(len(lidar_osm_building_levels_perc), lidar_osm_building_levels_perc)
print(len(lidar_osm_levels_perc), lidar_osm_levels_perc)
print(len(lidar_osm_random_fallback_perc), lidar_osm_random_fallback_perc)
print(len(overture_height_perc), overture_height_perc)
print(len(overture_num_floors_perc), overture_num_floors_perc)
print(len(overture_random_fallback_perc), overture_random_fallback_perc)

print(len(overall_mean_abs_diff), overall_mean_abs_diff)
print(len(overall_max_abs_diff), overall_max_abs_diff)
print(len(lidar_outside_explicit_overture_height), lidar_outside_explicit_overture_height)
```

```python
# function that takes in each stat array and a suitable title, prints the mean and plots the distribution
def print_and_plot_stats(stat_array, title):
    mean = np.mean(stat_array)
    median = np.median(stat_array)
    print(f"Mean of {title}: {mean:.2f}")
    print(f"Median of {title}: {median:.2f}")

    plt.hist(stat_array, bins=20, color='#80C080')
    plt.xlabel("Value")
    plt.ylabel("Frequency")
    plt.title(title)
    plt.show()
```

```python
mean = np.mean(relative_to_lidar_osm)
median = np.median(relative_to_lidar_osm)
print("How many times slower is Overture compared to LiDAR-OSM?")
print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")

plt.hist(relative_to_lidar_osm, bins=20, color='#80C080')
plt.xlabel("Relative Runtime")
plt.ylabel("Number of Occurrences")
plt.title("Histogram: Runtime of Overture Mode Relative to Runtime of LiDAR-OSM Mode")
plt.show()
```

```python
lidar_osm_breakdown_df = pd.DataFrame({
    "LiDAR": lidar_osm_hag_perc,
    "OSM Building:Height Tag": lidar_osm_building_height_perc,
    "OSM Height Tag": lidar_osm_height_perc,
    "OSM Building:Levels Tag * Multiplier": lidar_osm_building_levels_perc,
    "OSM Levels Tag * Multiplier": lidar_osm_levels_perc,
    "Random Fallback": lidar_osm_random_fallback_perc
})

sort_cols = [
    "LiDAR",
    "OSM Building:Height Tag",
    "OSM Height Tag",
    "OSM Building:Levels Tag * Multiplier",
    "OSM Levels Tag * Multiplier",
    "Random Fallback"
]

lidar_osm_breakdown_df = lidar_osm_breakdown_df.sort_values(
    by=sort_cols,
    ascending=False
)

fig, ax = plt.subplots(figsize=(20, 10))

lidar_osm_breakdown_df.plot(
    kind="bar",
    stacked=True,
    width=1,
    edgecolor="black",
    linewidth=0.6,
    ax=ax,
    color=["#80C080", "#84BCE1", "#AF93C3", "#F6BE80", "#E8DDEB", "#E68A88"]
)

ax.set_ylim(0, 100)
ax.set_ylabel("Percentage", fontsize=14)
ax.margins(x=0)

# Hide labels if there are many bars
ax.set_xticks([])

ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    fontsize=14
)

plt.title("Percentage of Buildings Derived from Each Data Source (LiDAR-OSM Mode)", size=16)
plt.show()
```

```python
print_and_plot_stats(lidar_osm_hag_perc, "Percentage of Building Heights Derived from LiDAR in LiDAR-OSM Mode")
```
